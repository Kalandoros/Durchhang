# Durchhang — Berechnungsprotokoll

Starte zuerst die Taipy-App (`Windeinfluss_mit_exakt.py`), führe im Browser eine Berechnung durch.
Die Werte werden automatisch aus `state.json` gelesen.

```python
import handcalcs.render
from math import sqrt, atan, degrees
import json

# Liest die aktuellen Werte aus state.json (geschrieben von der Taipy-App)
with open("state.json", "r", encoding="utf-8") as f:
    state = json.load(f)

# Alle Variablen aus state.json automatisch als lokale Variablen verfügbar machen
for key, value in state.items():
    globals()[key] = value

#print(f"Modus: {mode}")
print(f"F_G = {F_G} N/m,  F_W = {F_W} N/m")
print(f"sigma_0 = {sigma_0} N/mm²,  L = {span_L} m")
```

## 1. Windeinfluss

Die resultierende Last $F_{ges}$ aus Eigengewicht und Windlast nach Pythagoras.

```python
%%render
F_ges = sqrt(F_G**2 + F_W**2)
```

## 2. Ausschwingwinkel

Der Winkel $\alpha$, um den der Leiter durch den Wind ausgeschlagen wird.

```python
%%render
alpha = degrees(atan(F_G / F_W))
```

## Werte aktualisieren

Wenn du Werte in der Taipy-GUI (im Browser) änderst, führe diese Zelle aus, um die neuen Werte zu lesen, und dann die `%%render`-Zellen erneut.

```python
# Liest die aktualisierten Werte aus state.json
with open("state.json", "r", encoding="utf-8") as f:
    state = json.load(f)

# Alle Variablen automatisch aktualisieren
for key, value in state.items():
    globals()[key] = value
print(f"Aktualisiert: F_G = {F_G}, F_W = {F_W}")
print("Führe nun die %%render-Zellen erneut aus.")
```
